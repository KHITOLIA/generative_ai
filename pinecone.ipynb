```python
from dotenv import load_dotenv
from pinecone import Pinecone, ServerlessSpec
load_dotenv()
```

```python
import os
PINECONE_API_KEY = os.getenv("PINECONE_API_KEY")
```

```python
PINECONE_API_KEY
```

```python
pc = Pinecone(api_key = PINECONE_API_KEY)
```

```python
index_name = 'machine-learning'
if not pc.has_index(index_name):
    pc.create_index(
        name = index_name,
        dimension = 1024,
        metric = 'cosine', 
        spec = ServerlessSpec(cloud = 'aws', region = 'us-east-1')
    )
index = pc.Index(index_name)
```

```python
from langchain_pinecone import PineconeEmbeddings, PineconeVectorStore
```

```python
embedding_model = PineconeEmbeddings(model = 'multilingual-e5-large')
```

```python
vectorstore = PineconeVectorStore(
    index = index,
    embedding = embedding_model
)
```

```python
from langchain_core.documents import Document

document_1 = Document(
    page_content="I had chocolate chip pancakes and scrambled eggs for breakfast this morning.",
    metadata={"source": "tweet"},
    id=1,
)

document_2 = Document(
    page_content="The weather forecast for tomorrow is cloudy and overcast, with a high of 62 degrees.",
    metadata={"source": "news"},
    id=2,
)

document_3 = Document(
    page_content="Building an exciting new project with LangChain - come check it out!",
    metadata={"source": "tweet"},
    id=3,
)

document_4 = Document(
    page_content="Robbers broke into the city bank and stole $1 million in cash.",
    metadata={"source": "news"},
    id=4,
)

document_5 = Document(
    page_content="Wow! That was an amazing movie. I can't wait to see it again.",
    metadata={"source": "tweet"},
    id=5,
)

document_6 = Document(
    page_content="Is the new iPhone worth the price? Read this review to find out.",
    metadata={"source": "website"},
    id=6,
)

document_7 = Document(
    page_content="The top 10 soccer players in the world right now.",
    metadata={"source": "website"},
    id=7,
)

document_8 = Document(
    page_content="LangGraph is the best framework for building stateful, agentic applications!",
    metadata={"source": "tweet"},
    id=8,
)

document_9 = Document(
    page_content="The stock market is down 500 points today due to fears of a recession.",
    metadata={"source": "news"},
    id=9,
)

document_10 = Document(
    page_content="I have a bad feeling I am going to get deleted :(",
    metadata={"source": "tweet"},
    id=10,
)

documents = [
    document_1,
    document_2,
    document_3,
    document_4,
    document_5,
    document_6,
    document_7,
    document_8,
    document_9,
    document_10,
]
```

```python
documents
```

```python
vectorstore.add_documents(documents = documents, ids = [str(i) for i in range(len(documents))])
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs = {"k" : 3}
)
```

```python
query = input("Enter your query : ")

retriever.invoke(query)
```

```python
from langchain_pinecone import PineconeRerank
re_ranker = PineconeRerank()
```

```python
from langchain_classic.document_loaders import PyPDFLoader
loader = PyPDFLoader(file_path = r"C:\Users\Admin\Desktop\projects\Gen-AI_10_30_WD\data\Machine learning System Design.pdf")

docs = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size = 150,
    chunk_overlap = 30
)
chunks = splitter.split_documents(documents = docs)
```

```python
chunks
```

```python
vectorstore.add_documents(documents = chunks)
```

```python
query = input("Enter your question : ")
retriever = vectorstore.as_retriever(
    search_kwargs = {"k" : 20}
)
retrieved_docs = retriever.invoke(query)

re_ranked_docs = re_ranker.compress_documents(retrieved_docs, query)
print(retrieved_docs)
```

```python
len(re_ranked_docs)
content = ''
for i in re_ranked_docs:
    content += i.page_content
```
