```python
from langchain_core.tools import tool

@tool
def multiply(a : int ,b : int)-> int:
    '''Multiply two numbers a and b'''
    return a * b
```

```python
result = multiply.invoke({"a" : 10, "b" : 5})
```

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field

class Multiply_Input(BaseModel):
    a : int = Field(description= 'First no for multiplication', required = True)
    b : int = Field(description = 'Second no for multiplication', required = True)
```

```python
def multiply(a, b):
    return a * b
multiply_tool = StructuredTool.from_function(
    func = multiply,
    name = 'mulplication_tool',
    description = 'Multiplies two numbers',
    args_schema = Multiply_Input
)
```

```python
multiply_tool.args_schema.model_json_schema()
```

```python
multiply_tool.invoke({"a" : 2, "b" : 3})
```
