```python
from langchain_openai import ChatOpenAI
from langchain_classic.tools import tool
from langchain_core.messages import HumanMessage
from dotenv import load_dotenv
import requests
load_dotenv()
```

```python
@tool
def multiply(a :int, b : int) -> int:
    '''given 2 numbers a and b this tool return their product'''
    return a * b
```

```python
print(multiply.invoke({"a" : 3, "b" : 2}))
```

```python
llm = ChatOpenAI(model = 'gpt-5')
```

```python
llm.invoke("hi")
```

```python
llm_with_tools = llm.bind_tools([multiply])
```

```python
llm_with_tools.invoke("hi how are you?")
```

```python
query = HumanMessage(content = 'can you multiply 3 with 1000')
```

```python
messages = [query]
```

```python
messages
```

```python
result = llm_with_tools.invoke(messages)
```

```python
result
```

```python
messages.append(result)
```

```python
from langchain_core.tools import InjectedToolArg
from typing import Annotated
```

```python
@tool
def get_conversion_rate(base_currency : str, target_currency : str) -> float:
    '''This function fetches the currency conversion factor between a give base currency and target currecy'''
    url = f'https://v6.exchangerate-api.com/v6/c754eab14ffab33112e380ca/pair/{base_currency}/{target_currency}'
    response = requests.get(url)
    return response.json()
```

```python
get_conversion_rate.invoke({"base_currency" : "USD", "target_currency" : "INR"})
```

```python
@tool
def convert(base_currency_value : int, conversion_rate : Annotated[float, InjectedToolArg]) -> float:
    '''given a currency conversion rate this function can calculate the target currency value from a given base currency value'''
    return base_currency_value * conversion_rate
```

```python
convert.args
```

```python
get_conversion_rate.invoke({"base_currency" : "USD", "target_currency" : "INR"})
```

```python
convert.invoke({"base_currency_value" : 10, "conversion_rate" : 95.8689})
```

```python
llm = ChatOpenAI(model = 'gpt-5')
```

```python
llm_with_tools = llm.bind_tools([get_conversion_rate, convert])
```

```python
messages = [HumanMessage(content = 'What is the conversion factor between INR and USD and based on that convert 10 INR TO USD')]
```

```python
messages
```

```python
ai_message = llm_with_tools.invoke(messages)
```

```python
ai_message
```

```python
messages.append(ai_message)
```

```python
ai_message.tool_calls
```

```python
import json

for tool_call in ai_message.tool_calls:
    if tool_call['name'] == 'get_conversion_rate':
        tool_message1 = get_conversion_rate.invoke(tool_call)
        conversion_rate = json.loads(tool_message1.content)['conversion_rate']
        messages.append(tool_message1)
        tool_call['args']['conversion_rate'] = conversion_rate
        tool_message2 = convert.invoke(tool_call)
        messages.append(tool_message2)
```
